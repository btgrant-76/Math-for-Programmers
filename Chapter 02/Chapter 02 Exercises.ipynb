## 2.2 Plane vector arithmetic

```python
def add(v1, v2):
    return (v1[0] + v2[0], v1[1] + v2[1])
```

```python
first_vector = (4, 3)
second_vector = (-1, 1)
sum_vector = add(first_vector, second_vector)

draw(
    Points(first_vector, second_vector, sum_vector),
    Arrow(tip=first_vector, color=purple),
    Arrow(tip=second_vector, color=red),
    Arrow(tip=sum_vector, tail=first_vector, color=red), # this is tip-to-tail
    Arrow(tip=sum_vector, color=black)
)
```
